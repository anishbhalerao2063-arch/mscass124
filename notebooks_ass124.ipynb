```python
"""
Write a python code. Accept a paragraph from the user.
Perform the following tasks.
•	Count total words 
•	Count unique words 
•	Find the frequency of every word 
•	Display the most frequent word 
•	Display words appearing only once 
•	Sort words alphabetically 
"""
import string
from collections import Counter
para=input("enter")
#convert to lower_case 
para=para.lower()

#split para into words 
words=para.split()

print("total_words",len(words))
print("uni",len(set(words)))

freq=Counter(words)
print(freq)

if words:
    print("most",max(freq,key=freq.get))
else:
    print("no words")
print("Words appearing once:",
      [word for word in freq if freq[word] == 1])

# Sort words alphabetically
print("Alphabetical order:", sorted(words))

```

```python
"""
Write a Python program to create a function validate_customer() 
that validates customer information using the following rules:
1.     Customer ID must be unique.& #x20;
2.     Customer age must be 18 or above.&#x20;
3.     Customer must have a valid email address.&#x20;
4.     Customer must agree to the terms and conditions.&#x20;
Use **Set, Boolean values, operators, and functions** in the program
"""

customer_ids = {101, 102, 103}

def validate_customer(cid, age, email, agree):
    if cid not in customer_ids and age >= 18 and "@" in email and agree:
        customer_ids.add(cid)
        return True
    else:
        return False

cid = int(input("Enter customer ID: "))
age = int(input("Enter age: "))
email = input("Enter email: ")
agree = input("Agree to terms? (True/False): ") == "True"

print("Valid customer:", validate_customer(cid, age, email, agree))

```

```python
"""
(b) Write a Python program to create a function check_eligibility ()  
to determine whether a student is eligible to appear for an examination. 
The student will be eligible only if all the following conditions are satisfied:
•   Attendance must be at least 75%. 
•   Assignment must be submitted. 
•   Student must have paid the examination fee. 
•   Student must have completed the required practical work. 
•   Student must have cleared all previous dues.
The function should return a Boolean value (True or False) indicating whether the student is eligible for the examination.
"""

def check_eligibility(attendance, assignment, fee, practical, dues):
    if attendance >= 75 and assignment and fee and practical and dues:
        return True
    else:
        return False

attendance = float(input("Enter attendance percentage: "))
assignment = input("Assignment submitted? (True/False): ") == "True"
fee = input("Exam fee paid? (True/False): ") == "True"
practical = input("Practical completed? (True/False): ") == "True"
dues = input("Previous dues cleared? (True/False): ") == "True"

print("Eligible:", check_eligibility(attendance, assignment, fee, practical, dues))
```

```python
"""
Write a Python program using NumPy to sort an array in ascending 
and descending order and display the indices of the sorted elements using np.argsort
"""

import numpy as np 

import numpy as np

arr = np.array([33, 44, 5511, 111])

ascending = np.sort(arr)
print("ascending_array", ascending)

descending = np.sort(arr)[::-1]
print("descending", descending)

indices = np.argsort(arr)
print("Indices of sorted element", indices)

descending_order = np.argsort(arr)[::-1]
print("descending_order", descending_order)
```

```python
"""
) **Write a Python program using NumPy to&#x20;**&#x61;
accept 10 numbers from the user and create a NumPy array. 
Display all even numbers and odd numbers separately 
and find the second-largest and second-smallest elements .
"""


import numpy as np

# Accept 10 numbers from the user
numbers = []

for i in range(10):
    n = int(input("Enter a number: "))
    numbers.append(n)

# Create a NumPy array
arr = np.array(numbers)
print("NumPy Array:", arr)

# Display even numbers
even = arr[arr % 2 == 0]
print("Even numbers:", even)

# Display odd numbers
odd = arr[arr % 2 != 0]
print("Odd numbers:", odd)

# Find second-largest and second-smallest elements
unique_arr = np.unique(arr)
unique_arr = np.sort(unique_arr)

if len(unique_arr) >= 2:
    print("Second-smallest element:", unique_arr[1])
    print("Second-largest element:", unique_arr[-2])
else:
    print("Second-largest and second-smallest elements do not exist.")
```

```python
"""Write a Python program using NumPy to apply different discounts to an array of product prices:&#x20;

- 10%      discount for prices above ₹5000&#x20;
- 5%      discount for prices between ₹2000 and ₹5000&#x20;
- No    discount below ₹2000.
"""


import numpy as np

prices = np.array([1500, 2500, 4000, 5500, 8000])

discount = np.zeros(len(prices))

discount[prices > 5000] = 0.10
discount[(prices >= 2000) & (prices <= 5000)] = 0.05

final_prices = prices - (prices * discount)

print("Original prices:", prices)
print("Discount:", discount * 100, "%")
print("Final prices:", final_prices)
```

```python
"""
**)** Write a Python program using NumPy to analyze daily 
temperature data and find the number of days where the temperature 
was above the average temperature.
"""

import numpy as np

temp = np.array([30, 32, 28, 35, 31, 29, 36])

avg = np.mean(temp)

print("Average temperature:", avg)
print("Days above average:", np.sum(temp > avg))
```
